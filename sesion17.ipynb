 ## Negociación: una introducción práctica

 La negociación no consiste únicamente en conseguir que la otra persona acepte lo que nosotros queremos. Una buena negociación consiste en **entender qué necesita cada persona, descubrir qué está causando el conflicto y encontrar posibles soluciones**.

 Las ideas de negociación del FBI, la negociación de rehenes de Harvard y el trabajo de Chris Voss tienen un punto en común: **antes de intentar convencer, hay que escuchar y comprender**.

 ## 1\. Negociar es descubrir

 Una de las ideas más importantes es cambiar la mentalidad con la que entramos a una negociación.

 Una mentalidad tradicional sería:

 > "Tengo que convencer a esta persona para que acepte mi propuesta."

 La mentalidad de descubrimiento sería:

 > "Tengo que entender qué necesita esta persona y por qué está tomando esta posición."

 Esto es importante porque lo que una persona dice que quiere no siempre representa lo que realmente necesita.

 Por ejemplo:

 Una persona dice:

 > "Quiero que me pagues $2.000."

 Pero detrás de esa petición puede existir otra necesidad:

 > "Necesito recibir el dinero esta semana."

 Si solamente discutimos sobre los $2.000, podemos quedarnos atrapados en el precio. Si descubrimos la verdadera necesidad, pueden aparecer otras soluciones.

 Por eso, negociar es en gran medida un **proceso de descubrimiento**.

---

 ## 2\. La barrera del "NO"

 En una negociación, escuchar "no" no necesariamente significa que la conversación terminó.

 El "no" puede significar diferentes cosas:

 - La persona no está de acuerdo.
- No entiende la propuesta.
- No confía todavía.
- Se siente presionada.
- Quiere mantener el control.
- Necesita más información.
- La propuesta realmente no funciona para ella.

 Por eso, en lugar de intentar vencer inmediatamente el "no", podemos preguntarnos:

 > "¿Por qué esta persona está diciendo que no?"

 Por ejemplo:

 Empleado:

 > "No puedo entregar esto mañana."

 En lugar de responder:

 > "Necesito que lo hagas."

 Podemos preguntar:

 > "¿Qué hace difícil entregarlo mañana?"

 Ahora estamos intentando descubrir el problema.

 Quizás el empleado responda:

 > "Porque también tengo que terminar el informe del cliente."

 Ya tenemos información nueva. El problema quizá no sea falta de voluntad, sino exceso de trabajo.

 ## 3\. La técnica del reflejo

 El reflejo, o _mirroring_, es una técnica sencilla asociada con la metodología de Chris Voss.

 Consiste en repetir las últimas palabras importantes que acaba de decir la otra persona, normalmente entre una y tres palabras.

 Ejemplo:

 Persona:

 > "No puedo terminarlo esta semana porque tengo otros proyectos."

 Respuesta:

 > "¿Otros proyectos?"

 La persona probablemente continuará explicando:

 > "Sí, tengo dos proyectos que tienen prioridad porque los clientes los necesitan antes."

 El objetivo no es repetir mecánicamente. El objetivo es **estimular a la otra persona para que explique más**.

 Otro ejemplo:

 Cliente:

 > "Ese precio es demasiado alto."

 Vendedor:

 > "¿Demasiado alto?"

 Cliente:

 > "Sí, porque otro proveedor me ofrece algo similar por menos."

 Ahora el vendedor tiene información que antes no tenía.

 El principio es sencillo:

 > **Hablar menos puede ayudarte a descubrir más.**

 ## 4\. El poder de las preguntas

 Las preguntas son una de las herramientas más importantes de la negociación.

 Especialmente útiles son las preguntas abiertas que empiezan con:

 - ¿Cómo...?
- ¿Qué...?

 Estas preguntas permiten obtener información en lugar de limitar la conversación a un "sí" o un "no".

 Por ejemplo:

 > "¿Puedes bajar el precio?"

 La respuesta puede ser:

 > "No."

 En cambio:

 > "¿Cómo podemos hacer que esto funcione dentro de nuestro presupuesto?"

 La segunda pregunta obliga a pensar en posibilidades.

 Otro ejemplo:

 > "¿Puedes entregarlo mañana?"

 es una pregunta cerrada.

 Mientras que:

 > "¿Cómo podríamos organizar la entrega para que funcione para ambos?"

 abre una conversación diferente.

---

 ## 5\. Las preguntas calibradas

 Chris Voss utiliza el concepto de **preguntas calibradas** para referirse a preguntas diseñadas para obtener información y hacer que la otra persona participe en la búsqueda de una solución.

 Suelen comenzar con:

 > "¿Cómo...?"

 o

 > "¿Qué...?"

 Un ejemplo clásico sería:

 > "¿Cómo se supone que haga eso?"

 Esta pregunta no necesariamente significa "no quiero hacerlo".

 Puede significar:

 > "Explícame cómo podemos hacer esto teniendo en cuenta las limitaciones que existen."

 Por ejemplo:

 Jefe:

 > "Necesito que termines todo hoy."

 Empleado:

 > "¿Cómo se supone que haga eso teniendo estas otras tareas?"

 Esto permite que el jefe vea las limitaciones que quizá no había considerado.

 ## 6\. Las emociones son parte de la negociación

 Una negociación no es solamente lógica, números y argumentos.

 Las personas también tienen emociones.

 Una persona puede estar:

 - preocupada,
- frustrada,
- desconfiada,
- molesta,
- insegura,
- presionada.

 Si ignoramos esas emociones, podemos tener dificultades aunque nuestra propuesta sea lógica.

 Por eso es importante reconocer lo que la otra persona está experimentando.

 Por ejemplo:

 > "Parece que te preocupa que esto vuelva a generar el mismo problema."

 Esto no significa necesariamente estar de acuerdo con la persona.

 Significa demostrar que estamos intentando comprender su perspectiva.

 La diferencia es importante:

 **Validar una emoción no significa necesariamente aceptar la posición de la persona.**

 Puedes entender por qué alguien está preocupado y aun así no aceptar su propuesta.

 ## 7\. Negociación y empatía

 La empatía en negociación no significa ser demasiado complaciente.

 Significa intentar comprender:

 > "¿Cómo está viendo esta situación la otra persona?"

 Por ejemplo, un cliente puede decir:

 > "No quiero firmar porque ya tuve problemas con otra empresa."

 Una respuesta poco útil sería:

 > "Nosotros somos diferentes."

 Una respuesta orientada al descubrimiento sería:

 > "¿Qué ocurrió con la otra empresa?"

 Ahora podemos descubrir qué experiencia generó la desconfianza.

 Quizás el cliente diga:

 > "Me prometieron una fecha de entrega y terminaron retrasándose un mes."

 Entonces sabemos que su principal preocupación no es necesariamente el precio. Es el cumplimiento.

---

 ## 8\. Jefe y empleado: convertir una orden en una conversación

 La negociación también puede aplicarse a las relaciones laborales.

 Una estructura tradicional podría ser:

 > Jefe: "Haz esto."

 > Empleado: "Está bien."

 Pero cuando existe un problema, puede ser más productivo convertir la situación en una conversación.

 Por ejemplo:

 > "Tenemos que terminar este proyecto el viernes. ¿Qué necesitamos cambiar para conseguirlo?"

 El empleado puede responder:

 > "Si quitamos esta tarea de mi responsabilidad, puedo terminar el proyecto."

 Ahora aparece una posible solución.

 La diferencia es que la conversación pasó de:

 > "Hazlo porque yo lo digo."

 a:

 > "Tenemos un problema. Descubramos cómo resolverlo."

---

 ## 9\. "Never Split the Difference"

 El concepto central de _Never Split the Difference_, de Chris Voss, cuestiona la idea de que una negociación siempre debe terminar dividiendo la diferencia.

 Por ejemplo:

 Una persona quiere vender algo por $100.

 La otra quiere pagar $60.

 Una solución tradicional sería:

 > "Hagamos $80."

 Pero eso no necesariamente significa que $80 sea la mejor solución.

 Quizás el vendedor necesita vender rápidamente.

 Quizás el comprador puede pagar $70 hoy.

 Quizás existe una diferencia en las condiciones de entrega.

 Quizás se puede modificar la cantidad.

 La negociación no tiene que centrarse únicamente en el número.

 Hay que descubrir:

 > "¿Qué necesita cada persona?"

---

 ## 10\. Persuadir no es manipular

 Otro concepto importante es diferenciar persuasión de manipulación.

 La manipulación busca controlar a la otra persona ocultando información, explotando sus emociones o llevándola hacia algo que principalmente beneficia al manipulador.

 La persuasión puede ser diferente.

 Persuadir significa presentar una idea de manera que la otra persona pueda comprenderla y tomar una decisión.

 En una negociación ética, el objetivo debería ser:

 > **Entender las necesidades de la otra persona y explicar claramente las propias para encontrar una solución.**

 La empatía sirve para comprender, no para explotar.

 ## 11\. ¿Cómo se conectan todas estas técnicas?

 Podemos verlo como un proceso:

 ### Paso 1: Escuchar

 No intentar responder inmediatamente.

 ### Paso 2: Comprender

 Intentar descubrir qué está pensando y qué necesita la otra persona.

 ### Paso 3: Identificar las emociones

 Reconocer preocupaciones, frustraciones o inseguridades.

 ### Paso 4: Utilizar el reflejo

 Repetir algunas palabras importantes para estimular a la persona a explicar más.

 ### Paso 5: Hacer preguntas calibradas

 Utilizar preguntas como:

 > "¿Cómo podemos resolver esto?"

 > "¿Qué necesitaríamos cambiar?"

 > "¿Cómo se supone que podemos hacer eso?"

 ### Paso 6: Descubrir intereses

 Separar lo que la persona dice que quiere de lo que realmente necesita.

 ### Paso 7: Buscar alternativas

 Una vez comprendido el problema, buscar diferentes posibilidades.

 ### Paso 8: Llegar a un acuerdo

 El acuerdo aparece después del proceso de comprensión, no necesariamente al principio.

---

 ## 12\. La idea central

 Todos estos conceptos pueden resumirse en una sola idea:

 > **Negociar no es intentar ganar una discusión. Es intentar comprender el problema suficientemente bien como para encontrar una solución.**

 Por eso, una buena conversación de negociación podría seguir este esquema:

 > **Escuchar → comprender → preguntar → descubrir → proponer → colaborar.**

 Y esto conecta directamente con tu última pregunta:

 > **¿Qué pasaría si tratáramos cada conversación como una colaboración?**

 En lugar de pensar:

 > "Yo contra la otra persona."

 pensamos:

 > "Los dos tenemos un problema que necesitamos entender y resolver."

 Esa es probablemente la mentalidad más importante detrás de todas estas técnicas.

PNL: un modelo para entender cómo pensamos y actuamos
La Programación Neurolingüística, conocida como PNL, puede entenderse como un conjunto de ideas y técnicas que intenta estudiar cómo las personas procesan sus experiencias, utilizan el lenguaje y desarrollan determinados patrones de pensamiento y comportamiento.
Una manera sencilla de entender la propuesta es imaginar que nuestra mente tiene una especie de "manual de usuario".
No significa que exista literalmente un manual dentro del cerebro. La idea es utilizar modelos que nos permitan observar:
"¿Cómo estoy pensando, interpretando y reaccionando ante lo que me ocurre?"

1. El "manual de usuario" de la propia mente
Todos tenemos formas habituales de:
interpretar lo que sucede,


prestar atención,


recordar experiencias,


hablar con nosotros mismos,


tomar decisiones,


reaccionar emocionalmente,


comunicarnos con otras personas.


Muchas veces hacemos estas cosas automáticamente.
Por ejemplo, dos personas pueden recibir exactamente la misma crítica.
Una puede pensar:
"Me están atacando."
Otra puede pensar:
"Me están dando información que puedo utilizar para mejorar."
El acontecimiento es similar, pero la interpretación puede ser diferente.
La PNL intenta estudiar precisamente este tipo de procesos.
La pregunta sería:
¿Qué ocurre entre lo que sucede y la respuesta que damos?

2. La "caja negra" de la mente
Podemos imaginar la mente como una caja negra.
Tenemos:
Entrada → proceso interno → respuesta
Por ejemplo:
Entrada:
 Un jefe dice: "Tenemos que hablar."
Proceso interno:
 La persona interpreta: "Seguro hice algo mal."
Respuesta:
 Se pone nerviosa.
Otra persona podría interpretar:
"Probablemente quiere revisar un proyecto."
Y su reacción sería diferente.
La PNL intenta prestar atención a ese proceso intermedio.
En lugar de estudiar solamente:
"¿Qué pasó?"
se pregunta:
"¿Cómo interpretó la persona lo que pasó?"

3. ¿Qué significa "neurolingüística"?
El término puede dividirse en tres partes:
Neuro
Hace referencia a los procesos relacionados con el sistema nervioso y la experiencia interna.
La pregunta sería:
"¿Cómo procesa una persona su experiencia?"
Lingüística
Hace referencia al lenguaje.
El lenguaje es importante porque utilizamos palabras para describir, interpretar y comunicar nuestras experiencias.
Por ejemplo:
"Esto es imposible."
no genera la misma conversación que:
"¿Qué tendría que cambiar para hacerlo posible?"
Programación
En PNL, "programación" se utiliza para describir patrones organizados de pensamiento y comportamiento.
La idea es que podemos desarrollar determinados patrones y, en algunos casos, modificarlos.
Es importante aclarar que esto no significa que podamos programar literalmente el cerebro como si fuera un ordenador.

4. La idea de ingeniería inversa
Una de las ideas interesantes de la PNL es utilizar una especie de ingeniería inversa del comportamiento.
La ingeniería inversa parte de algo que funciona y trata de descubrir cómo funciona.
Por ejemplo, si tienes una máquina que funciona perfectamente, puedes desmontarla para estudiar:
"¿Qué piezas tiene y cómo interactúan?"
Aplicado al comportamiento humano, la pregunta sería:
"¿Qué hace exactamente esta persona que consigue este resultado?"
Esto cambia el enfoque.
En lugar de preguntar solamente:
"¿Por qué esta persona tiene este problema?"
se intenta preguntar:
"¿Cómo está funcionando este patrón?"

5. "¿Por qué?" frente a "¿Cómo?"
Esta diferencia aparece constantemente en el material que estás estudiando.
Una pregunta orientada al "por qué" sería:
"¿Por qué tienes miedo de hablar en público?"
Una pregunta orientada al "cómo" sería:
"¿Qué ocurre exactamente en tu mente cuando tienes que hablar delante de otras personas?"
La segunda pregunta intenta identificar el proceso.
Quizás la persona diga:
"Antes de hablar empiezo a imaginar que todos se van a dar cuenta de que estoy nervioso."
Ahora tenemos más información sobre el proceso.
Podemos continuar preguntando:
"¿Qué piensas inmediatamente después?"
"¿Qué haces con tu cuerpo?"
"¿Qué te dices internamente?"
"¿Qué ocurre después?"
De esta manera se intenta descomponer una experiencia en partes.

6. El modelado
El concepto de modelado es uno de los elementos más importantes de la PNL.
La idea es estudiar a una persona que obtiene buenos resultados en determinada actividad y tratar de identificar los patrones que utiliza.
Por ejemplo, imaginemos a un vendedor excepcional.
En lugar de decir simplemente:
"Es muy bueno vendiendo."
se intenta observar:
¿Cómo inicia una conversación?


¿Qué preguntas hace?


¿Cómo escucha?


¿Cómo explica el producto?


¿Cómo responde a las objeciones?


¿Qué palabras utiliza?


¿Cómo utiliza su tono de voz?


¿Cómo estructura una negociación?


¿Qué hace cuando el cliente dice que no?


El objetivo es pasar de:
"Esta persona es buena."
a:
"¿Qué hace exactamente esta persona?"

7. El origen de la idea de estudiar a personas excepcionales
La PNL surgió en la década de 1970 alrededor del trabajo de Richard Bandler y John Grinder, quienes estudiaron patrones utilizados por terapeutas destacados, entre ellos Fritz Perls, Virginia Satir y Milton Erickson.
Su intención era identificar estructuras y patrones de comunicación que pudieran ser descritos y enseñados.
La idea general era:
"Si una persona consigue resultados excepcionales, estudiemos qué hace de manera diferente."
Esto es lo que posteriormente se conoce como modelado.
Aquí es importante diferenciar entre la historia de la PNL y la evidencia científica posterior: que una técnica haya sido desarrollada a partir de observar a personas exitosas no demuestra por sí mismo que el modelo funcione científicamente.

8. Deconstruir una habilidad
Imagina que quieres aprender a negociar como una persona que es muy buena negociando.
En lugar de intentar copiarla superficialmente, puedes dividir su comportamiento.
Comunicación
¿Cómo empieza las conversaciones?
Preguntas
¿Qué preguntas utiliza?
Escucha
¿Cuánto habla y cuánto escucha?
Emociones
¿Cómo responde cuando la otra persona está molesta?
Lenguaje
¿Qué palabras utiliza?
Decisiones
¿Cómo decide cuándo avanzar y cuándo esperar?
Comportamiento
¿Qué hace concretamente?
Esto convierte una habilidad que parece "natural" en una serie de comportamientos que pueden estudiarse.

9. "Programar" el éxito
Cuando el material habla de "programar tu éxito", normalmente se refiere a desarrollar deliberadamente determinados patrones.
Por ejemplo:
Una persona puede tener el hábito de pensar:
"No soy bueno hablando con clientes."
Podría intentar cambiar la pregunta hacia:
"¿Qué habilidad específica necesito desarrollar para comunicarme mejor con los clientes?"
El segundo enfoque convierte una afirmación general sobre la identidad en un problema específico que puede trabajarse.
Sin embargo, es importante no entender esto como:
"Pensar positivamente garantiza el éxito."
La realidad es más compleja. El rendimiento depende también de conocimientos, habilidades, experiencia, contexto, oportunidades y muchos otros factores.

10. Aplicación al liderazgo
Estas ideas pueden aplicarse al liderazgo porque un líder necesita comprender cómo las personas interpretan las situaciones.
Por ejemplo, ante un error de un empleado:
Enfoque poco útil:
"¿Por qué hiciste esto mal?"
Enfoque orientado al proceso:
"¿Qué ocurrió?"
"¿Cómo tomaste esa decisión?"
"¿Qué información tenías en ese momento?"
"¿Qué podríamos cambiar para evitar que vuelva a ocurrir?"
La segunda conversación permite estudiar el proceso que produjo el resultado.
Eso puede ser más útil para aprender del error.

11. Aplicación a las ventas
En ventas, el concepto de modelado puede utilizarse para estudiar a vendedores con buenos resultados.
Por ejemplo:
Un vendedor obtiene buenos resultados.
En lugar de decir:
"Tiene talento."
se puede estudiar:
"¿Qué hace durante una llamada?"
Quizás descubrimos que:
Hace preguntas antes de presentar el producto.


Identifica el problema del cliente.


Escucha las respuestas.


Personaliza la explicación.


Responde a las objeciones.


Confirma si la solución tiene sentido.


Pregunta por el siguiente paso.


Ahora tenemos un proceso que podemos estudiar y entrenar.

12. Aplicación a la negociación
La conexión con lo que estudiabas anteriormente es bastante directa.
La negociación pregunta:
"¿Qué necesita la otra persona?"
La PNL puede aportar una pregunta adicional:
"¿Cómo está interpretando esta persona la situación?"
Por ejemplo:
Cliente:
"No confío en este producto."
En lugar de responder inmediatamente:
"Sí puedes confiar."
podemos preguntar:
"¿Qué tendría que suceder para que te sintieras cómodo con la compra?"
Eso permite descubrir cómo está construyendo su percepción del problema.

13. La idea de "decodificar" una habilidad
La palabra "decodificar" puede entenderse como descomponer algo complejo en elementos observables.
Por ejemplo:
"Esta persona es un excelente líder."
Es una afirmación demasiado general.
La podemos descomponer:
¿Cómo comunica?
¿Cómo toma decisiones?
¿Cómo da retroalimentación?
¿Cómo maneja conflictos?
¿Cómo establece objetivos?
¿Cómo escucha?
¿Cómo responde bajo presión?
Así pasamos de una característica abstracta a comportamientos concretos.

14. Una advertencia importante sobre la PNL
Aquí hay una distinción importante para estudiar el tema correctamente.
La PNL utiliza términos relacionados con la neurología, el lenguaje y la psicología, pero eso no significa que todas sus afirmaciones sobre el cerebro estén científicamente demostradas.
Algunas técnicas de comunicación o aprendizaje pueden resultar útiles como herramientas prácticas, pero muchas afirmaciones específicas de la PNL han sido cuestionadas o no cuentan con evidencia científica sólida.
Por eso es mejor entenderla como:
Un conjunto de modelos y técnicas para analizar comunicación, pensamiento y comportamiento.
Y no como:
Un manual científicamente comprobado de cómo funciona el cerebro humano.

Resumen final
La idea central podría resumirse así:
La PNL intenta estudiar cómo las personas construyen sus experiencias internas, cómo utilizan el lenguaje y cómo esos patrones se relacionan con su comportamiento.
El concepto de ingeniería inversa busca descubrir cómo funciona un comportamiento.
El modelado busca estudiar a personas que tienen determinadas habilidades para identificar patrones que puedan ser aprendidos.
La aplicación práctica consiste en utilizar ese análisis para trabajar habilidades como:
comunicación,


liderazgo,


ventas,


negociación,


gestión de equipos,


aprendizaje,


resolución de problemas.


Y la pregunta que resume todo sería:
"En lugar de preguntarme solamente qué resultado obtuvo esta persona, ¿puedo entender qué proceso utilizó para obtenerlo?"
